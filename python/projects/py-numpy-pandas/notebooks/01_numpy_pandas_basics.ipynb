## 4. pandas: load and inspect data

```python
sales = pd.read_csv(DATA_DIR / "sales.csv", parse_dates=["date"])
sales.head()
```

```python
sales.info()
sales.describe(include="all")
```

## 5. pandas: select, filter, assign

```python
sales = sales.assign(
    gross_revenue=sales["units"] * sales["unit_price"],
    net_revenue=lambda df: df["gross_revenue"] * (1 - df["discount"]),
)
sales.loc[sales["channel"].eq("Online"), ["date", "region", "product", "net_revenue"]]
```

## 6. pandas: groupby and pivot tables

```python
region_summary = (
    sales.groupby("region", as_index=False)
    .agg(
        orders=("date", "count"),
        units=("units", "sum"),
        revenue=("net_revenue", "sum"),
    )
    .sort_values("revenue", ascending=False)
)
region_summary
```

```python
pd.pivot_table(
    sales,
    index="region",
    columns="product",
    values="net_revenue",
    aggfunc="sum",
    fill_value=0,
    margins=True,
)
```

## 7. Export results

```python
region_summary.to_csv(DATA_DIR / "region_summary.csv", index=False)
region_summary.to_parquet(DATA_DIR / "region_summary.parquet", index=False)
region_summary
```
